# RQ2 completed-artifact reader
This notebook contains no extraction, fitting, calibration, evaluation or statistical source logic. It only reads artifacts produced by the Python pipeline.

```python
from pathlib import Path
import json
import pandas as pd

cwd = Path.cwd().resolve()
candidates = [cwd, cwd / 'experiments2.0' / 'RQ2', cwd.parent]
rq2_root = next(path for path in candidates if (path / 'configs' / 'rq2.yaml').is_file())
output = rq2_root / 'outputs'  # change to outputs/mini_e2e for diagnostics
metrics = json.loads((output / 'metrics.json').read_text(encoding='utf-8'))
main = pd.read_csv(output / 'Table_RQ2_main.csv')
primary = pd.read_csv(output / 'Table_RQ2_primary_inference.csv')
metrics['evidence_status'], main, primary
```
